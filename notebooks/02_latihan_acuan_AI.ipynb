```python
from pathlib import Path
import os, sys, json, subprocess, hashlib

print('Mulai cell 1: memeriksa Python dan mengambil kode; belum memasang paket/model.', flush=True)

def run_git(arguments, stage, timeout=120, capture=False):
    print(stage + f' (batas {timeout} detik)', flush=True)
    git_env = os.environ.copy()
    git_env['GIT_TERMINAL_PROMPT'] = '0'
    command = ['git', '-c', 'http.lowSpeedLimit=1', '-c', 'http.lowSpeedTime=30', *arguments]
    try:
        result = subprocess.run(command, check=True, timeout=timeout, env=git_env,
                                stdin=subprocess.DEVNULL, text=True,
                                stdout=subprocess.PIPE if capture else None,
                                stderr=subprocess.STDOUT if capture else None)
    except subprocess.TimeoutExpired as exc:
        raise RuntimeError(stage + ' melewati batas waktu. Cell dihentikan; setup belum berjalan. '
                           'Periksa koneksi runtime/GitHub dan jalankan ulang cell ini. '
                           'Folder/data yang sudah ada tidak dihapus.') from exc
    except subprocess.CalledProcessError as exc:
        detail = (exc.stdout or '')[-2000:] if capture else 'Lihat pesan Git di atas.'
        raise RuntimeError(stage + ' gagal. ' + detail +
                           ' Tidak menghapus folder atau menimpa perubahan pengguna.') from exc
    return result.stdout.strip() if capture else None

INPUT_FORMAT = "xlsx" #@param ["csv", "xlsx"]
EXCEL_FILENAME = "Review_Esai_AI_Lengkap.xlsx" #@param {type:"string"}
EXCEL_SHEET = "Untuk_CSV_Guru" # Tab data lengkap; jangan ganti menjadi Ringkasan_AI.
TRACE_SCENARIO = "S3" #@param ["S1", "S2", "S3"]
TRACE_TRAIT = "isi" #@param ["isi", "organisasi", "kebahasaan", "mekanik"]

MODE = "ai_learning" #@param ["inspect", "ai_learning", "learning", "proposal"]
RUN_ENCODERS = True #@param {type:"boolean"}
RUN_TESTS = True #@param {type:"boolean"}
RUN_AC2_REFERENCE = True #@param {type:"boolean"}

if sys.version_info[:2] != (3, 12):
    raise RuntimeError("Pilih Runtime version 2026.07, Python 3.12, CPU sebelum setup.")
IN_COLAB = Path('/content').is_dir()
if os.environ.get('AES_NOTEBOOK_ROOT'):
    ROOT = Path(os.environ['AES_NOTEBOOK_ROOT']).resolve()
elif IN_COLAB:
    ROOT = Path('/content/Automated_Essay_Scoring')
else:
    ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents]
                 if (p/'configs/default.json').is_file()), None)
    if ROOT is None:
        raise FileNotFoundError('Buka notebook dari checkout proyek atau atur AES_NOTEBOOK_ROOT.')

if IN_COLAB and not ROOT.exists():
    run_git(['clone', '--depth', '1', '--branch', 'codex/aes-learning', '--single-branch',
             'https://github.com/MalikHariyanto/Automated_Essay_Scoring.git', str(ROOT)],
            'Mengambil source dari GitHub', timeout=120)
elif IN_COLAB:
    dirty = run_git(['-C', str(ROOT), 'status', '--porcelain', '--untracked-files=no'],
                    'Memeriksa perubahan lokal', timeout=20, capture=True)
    if dirty:
        raise RuntimeError('Checkout memiliki perubahan tracked. Simpan perubahan tersebut sebelum memperbarui kode.')
    branch = run_git(['-C', str(ROOT), 'branch', '--show-current'],
                     'Memeriksa branch aktif', timeout=20, capture=True)
    if branch != 'codex/aes-learning':
        raise RuntimeError('Checkout memakai branch lain; gunakan branch codex/aes-learning.')
    run_git(['-C', str(ROOT), 'fetch', 'origin', 'codex/aes-learning'],
            'Mengambil pembaruan dari GitHub', timeout=120)
    run_git(['-C', str(ROOT), 'merge', '--ff-only', 'FETCH_HEAD'],
            'Memperbarui checkout lokal', timeout=30)

if not (ROOT/'scripts/setup_notebook.py').is_file():
    raise FileNotFoundError('Checkout belum memuat setup notebook terbaru.')
os.chdir(ROOT)
OUTPUT = ROOT/'outputs'
OUTPUT.mkdir(exist_ok=True)
ENV_DIR = OUTPUT/'notebook_env'
AES_PYTHON = ENV_DIR/'bin/python'
WORKER_ENV = os.environ.copy()
WORKER_ENV.pop('PYTHONPATH', None)
WORKER_ENV.pop('PYTHONHOME', None)
WORKER_ENV.update(HF_HOME=os.environ.get('HF_HOME', str(OUTPUT/'cache/huggingface')), OMP_NUM_THREADS='2', MKL_NUM_THREADS='2')
print('Python notebook:', sys.version.split()[0])
print('Folder proyek:', ROOT)
print('Mode:', MODE, '| Model asli:', RUN_ENCODERS, '| Unit test:', RUN_TESTS)
print('Input:', INPUT_FORMAT, '| Excel:', EXCEL_FILENAME, '| Tab:', EXCEL_SHEET)
print('CELL 1 SELESAI — lanjut ke setup Python AES.', flush=True)
```

## 7. Tokenisasi, window, dan embedding

**Token IDs** adalah nomor potongan teks dalam kamus tokenizer. **Embedding** adalah vektor angka yang mewakili teks. BGE-M3 memakai maksimum 1024 token termasuk special tokens dan melaporkan bagian terpotong. RoBERTa memakai semua teks melalui **window**, potongan maksimum 512 token termasuk special tokens, dengan overlap 128 token isi.

Untuk checkpoint RoBERTa ini tokenizer slow menghitung dua special tokens; kapasitas isi 510 dan langkah antarwindow 382. Mapping chunk ke esai menjaga agar output tetap satu baris per esai.

Default `RUN_ENCODERS=True` menjalankan checkpoint asli. Jika dipilih False, cell menjelaskan bahwa jalur ini dilewati. Unduhan checkpoint sekitar 2,8 GB. Gangguan jaringan dilaporkan; error implementasi tetap muncul.
Tambahan alur belajar menampilkan pasangan potongan token/ID, jumlah window, mapping chunk ke esai, dan contoh elemen vektor aktual dari esai pertama. Simbol tokenizer dapat menandai spasi atau potongan kata; ID adalah indeks kamus, bukan nilai esai. Vektor tidak memiliki satu arti kata per posisi yang bisa dibaca langsung.

## 8. Bentuk data dan forward UNTRAINED

**Shape** adalah ukuran susunan angka. Untuk satu esai: BGE `[1,1024]`, RoBERTa `[1,768]`, proyeksi BGE `[1,256]`, gabungan `[1,1024]`, dan output `[1,4]`. Empat kolom output selalu berurutan isi, organisasi, kebahasaan, mekanik.

**Forward** ialah perhitungan teks sampai output. Head baru **UNTRAINED**: belum belajar dari skor guru. Angkanya hanya diagnostik, bisa di luar 1–5, dan tidak boleh dipakai menilai siswa. Shape aktual hanya ditampilkan setelah checkpoint benar-benar dijalankan.

S1 menunjukkan RoBERTa → empat keluaran; S2 BGE → MLP → empat keluaran; S3 menunjukkan proyeksi dan gabungan → empat keluaran. Tabel diagnostik memuat empat trait setiap skenario. Kategori dihasilkan dari salinan angka kontinu melalui clipping dan kategori terdekat, bukan skor guru. Ketika head belum terlatih, kategori yang semuanya 1 tidak membuktikan tulisan siswa buruk.
Urutan algoritmenya: S1 memakai token RoBERTa → CLS tiap window → rata-rata per esai → head empat output; S2 memakai token BGE → dense L2 → MLP → empat output; S3 memakai kedua jalur → proyeksi BGE → concat → head. Concat menaruh RoBERTa di posisi 0–767 dan proyeksi BGE di posisi 768–1023. Tampilan contoh kedua bagian vektor memeriksa bahwa gabungannya tidak tertukar. Mean window bukan mean skor trait.


```python
if model_result is not None:
    diagnostic_rows = []
    for scenario, detail in model_result['scenarios'].items():
        print(scenario, '| SHAPES AKTUAL:', detail['shapes'])
        print('Contoh lima elemen vektor:', detail['vector_first_five'])
        for trait in detail['trait_order']:
            diagnostic_rows.append({'Skenario':scenario, 'Trait':trait,
                'Kontinu UNTRAINED':detail['continuous'][trait],
                'Kategori diagnostik':detail['categories'][trait], 'Status':'UNTRAINED'})
    show_table(['Skenario','Trait','Kontinu UNTRAINED','Kategori diagnostik','Status'], diagnostic_rows)
    print('Frozen/trainable aktual:', model_result['gradient_policy'])
    print('Angka di atas hanya diagnostik. Skor guru tidak diisi dari output tersebut.')
else:
    print('Shape aktual belum dihitung; lihat status akses checkpoint/pilihan encoder.')
if model_result is not None:
    for scenario, detail in model_result['scenarios'].items():
        print('ALUR', scenario, '— checkpoint encoder asli; head belum dilatih dari skor guru.')
        stages = []
        if scenario != 'S2':
            stages += [{'Tahap':'CLS setiap window', 'Shape':detail['shapes']['roberta_window_cls'],
                        'Contoh angka':detail['cls_per_window_first_five']},
                       {'Tahap':'Mean window per esai', 'Shape':detail['shapes']['roberta'],
                        'Contoh angka':detail['vector_first_five']['roberta']}]
        if scenario != 'S1':
            stages += [{'Tahap':'BGE dense L2', 'Shape':detail['shapes']['bge'],
                        'Contoh angka':detail['vector_first_five']['bge']}]
        if scenario == 'S2':
            stages += [{'Tahap':'MLP: Linear → GELU → dropout', 'Shape':detail['shapes']['mlp_hidden'],
                        'Contoh angka':detail['vector_first_five']['mlp_hidden']}]
        if scenario == 'S3':
            stages += [{'Tahap':'BGE Linear → LayerNorm', 'Shape':detail['shapes']['projection'],
                        'Contoh angka':detail['vector_first_five']['projection']},
                       {'Tahap':'Concat [RoBERTa; proyeksi BGE]', 'Shape':detail['shapes']['concat'],
                        'Contoh angka':detail['concat_segments']}]
        stages += [{'Tahap':'Head linear: isi, organisasi, kebahasaan, mekanik',
                    'Shape':detail['shapes']['output'], 'Contoh angka':detail['continuous']}]
        show_table(['Tahap', 'Shape', 'Contoh angka'], stages)
    print('Contoh vektor hanya lima posisi awal; pada concat kedua sumber juga ditampilkan.')
```

## 9. Frozen, trainable, dan pengujian teknis

**Frozen** berarti parameter BGE tidak diperbarui. **Trainable** berarti parameter RoBERTa, proyeksi S3, dan head dapat diperbarui saat training. **Gradien** menunjukkan perubahan parameter yang diperlukan untuk mengurangi kesalahan. Empat trait berbagi encoder dan loss; kesalahannya tidak otomatis independen.

**Unit test** memeriksa perilaku bagian kode: urutan target, window, bentuk `[B,4]`, loss yang menolak broadcasting, gradien model tiruan, reload, dan metrik. Fixture ialah input kecil buatan khusus untuk test, bukan dataset penelitian. Lulus unit test bukan bukti akurasi AES.

Default `RUN_TESTS=True` dan `RUN_AC2_REFERENCE=True`. Colab menyiapkan R/irrCAC jika diperlukan. Tabel berikut membandingkan **AC2 Python dan R pada lima fixture unit test**, termasuk kategori tak teramati. Selisih kecil diverifikasi dengan toleransi 5,01×10⁻⁶ karena output R dibulatkan lima desimal. Ini pemeriksaan rumus, bukan nilai AC2 esai penelitian. Kegagalan dependency R tetap dilaporkan.

```python
if RUN_AC2_REFERENCE:
    if IN_COLAB:
        subprocess.run(['bash', str(ROOT/'scripts/setup_colab.sh')],
                       cwd=ROOT, env=WORKER_ENV, check=True)
    else:
        run_aes([ROOT/'scripts/verify_ac2.py', '--require-r'])
    ac2_fixture_results = json.loads((OUTPUT/'ac2_verification.json').read_text())
    show_table(['Fixture unit test','AC2 Python','AC2 R/irrCAC','Selisih absolut'],
        [{'Fixture unit test':i+1,'AC2 Python':row['python'],
          'AC2 R/irrCAC':row['irrCAC_AC2'],'Selisih absolut':row['absolute_error']}
         for i,row in enumerate(ac2_fixture_results)])
    print('REFERENSI AC2 LULUS — fixture unit test, bukan akurasi AES.')
else:
    print('Referensi AC2/R belum dijalankan dalam sesi ini.')
if RUN_TESTS:
    run_aes(['-m', 'pytest', '-q', '-rA'])
    print('Unit test selesai; baca passed/failed/skipped pada hasil runner.')
else:
    print('Unit test belum dipilih pada putaran ini.')
```

## 10. Langkah supervised setelah skor guru tersedia

**Training supervised** memperbarui parameter dari pasangan teks dan skor guru. Pemeriksaan berikut selalu berjalan. Bila data belum layak, hanya tahap supervised yang dilewati dengan daftar kebutuhan konkret; inspeksi tetap selesai.

Syaratnya mencakup empat skor sah, transkripsi lengkap/verified, rubrik disahkan guru, identitas dan hubungan duplikasi diperiksa, serta review terhadap hash CSV terkini. Jangan mengubah flag agar training lolos.

`MODE="learning"` menjalankan latihan singkat yang dipisahkan dari eksperimen proposal. `MODE="proposal"` menyediakan S1/S2/S3, tuning, repeated CV, refit, dan final test. Pilih mode tersebut setelah data dan sumber daya benar-benar siap. Default inspect tidak menghitung loss atau metrik penelitian dari label kosong.
`MODE="ai_learning"` adalah latihan yang secara eksplisit memakai usulan AI. Ia membaca empat angka dari Ringkasan_AI, memakai teks asli yang tidak terpotong, menjaga group kemiripan usulan tetap utuh, lalu menjalankan S1/S2/S3 selama dua epoch tetap. Contoh split 50/50 dengan seed 2026 hanya untuk latihan kecil; rencana penelitian 80/20 dan repeated CV tetap terpisah. Angka metrik mengukur kecocokan dengan AI, bukan guru. Mode ini tidak mengubah score_* guru, confirmed, ready_for_training atau identitas.

## 11. Tabel empat metrik: hasil penelitian

Tabel ini selalu menampilkan **12 baris**: S1/S2/S3 × isi/organisasi/kebahasaan/mekanik. **RMSE** mengukur besar kesalahan dalam unit skor 1–5; lebih kecil lebih baik. **Pearson** mengukur hubungan linear, bukan kesepakatan. **QWK** dan **Gwet AC2** mengukur kesepakatan kategori dengan bobot kuadratik; lebih besar menunjukkan kesepakatan lebih kuat.

Untuk RMSE/Pearson, prediksi kontinu tidak di-clip atau dibulatkan. Untuk QWK/AC2, salinan prediksi dipetakan ke kategori JSON [1,2,3,4,5]. Titik tengah memilih kategori lebih rendah. Setiap trait dihitung sendiri, tidak diratakan menjadi satu target.

Dengan data draf sekarang, tanda **—** berarti **belum tersedia**, bukan nol. Nilai hanya muncul sesudah mode proposal menyelesaikan training, mengunci ketiga konfigurasi/refit, dan mengevaluasi test yang dipisahkan secara sah. Hasil latihan learning tidak dicampur dengan hasil final proposal. Tabel membaca prediksi test dan menghitung ulang metrik untuk memastikan cocok dengan sumber/manifest.

Contoh maknanya: untuk RMSE isi, setiap esai memerlukan pasangan skor isi guru dan skor isi model. Saat skor guru kosong, selisihnya tidak diketahui; program tidak boleh menciptakan nilai RMSE. Nilai AC2 pada fixture di cell sebelumnya juga tidak boleh dimasukkan ke tabel penelitian ini.
Untuk belajar, proses perhitungan juga ditampilkan setelah pasangan guru/prediksi test sah tersedia: cocokkan essay_id → lihat prediksi kontinu dan selisih → hitung RMSE/Pearson → clip dan konversi kategori pada salinan → hitung QWK/AC2 → tabel akhir. Pilih `TRACE_SCENARIO` dan `TRACE_TRAIT` di cell pertama. Ini memakai hasil test yang sama dengan tabel, tidak menciptakan label atau angka contoh sebagai pengganti penelitian.

Dalam mode ai_learning, tabel dan walkthrough memakai acuan AI serta validation dari split latihan. Contoh Pearson tinggi masih dapat disertai RMSE besar karena korelasi bukan agreement. Metrik undefined tetap dijelaskan, tidak diganti nol. Hasil ini tidak masuk tabel penelitian, tuning proposal, atau klaim keunggulan fusion.


```python
run_aes([ROOT/'scripts/notebook_step.py', 'report', '--config', CONFIG_PATH, '--mode', MODE])
report_file = 'ai_learning_report.json' if MODE == 'ai_learning' else 'final_learning_report.json'
final_report = json.loads((OUTPUT/report_file).read_text())
ai_scope = MODE == 'ai_learning'
print('EVALUASI TERHADAP ACUAN AI — LATIHAN, BUKAN GURU' if ai_scope else 'EVALUASI PENELITIAN TERHADAP GURU')
if ai_scope:
    for s,flow in final_report['trained_flows'].items():
        print(s, 'FORWARD SETELAH LATIHAN AI:', flow['shapes'])
        print('Contoh lima elemen vektor:', flow['vector_first_five'])
        print('Empat output kontinu:', flow['continuous'], '| kategori:', flow['categories'])
result_rows = [{'Skenario':row['scenario'],'Trait':row['trait'],
                **{metric:row[metric] for metric in ['RMSE','Pearson','QWK','AC2']},
                'Status':'Latihan — acuan AI' if ai_scope else ('Hasil test final' if row['status']=='available_final_test' else 'Belum tersedia')}
               for row in final_report['rows']]
show_table(['Skenario','Trait','RMSE','Pearson','QWK','AC2','Status'], result_rows)
print(final_report['note'])
if final_report.get('data_blockers'):
    print('Kebutuhan konkret sebelum nilai evaluasi dapat dihitung:')
    for reason in final_report['data_blockers']:
        print('-', reason)
for row in final_report['rows']:
    if ai_scope or row['status']=='available_final_test':
        print(row['scenario'], row['trait'], '| Proporsi prediksi di luar 1–5:', row['out_of_range_fraction'])
    if ai_scope and row['undefined_reasons']:
        print(row['scenario'], row['trait'], '| Metrik undefined:', row['undefined_reasons'])
    if row['status']=='available_final_test' and row['reason']:
        print(row['scenario'], row['trait'], '| Metrik undefined:', row['reason'])
print('Laporan lokal:', OUTPUT/report_file)
print('Alur belajar selesai. Output UNTRAINED, test teknis, dan hasil penelitian ditampilkan terpisah.')
print('LANGKAH PERHITUNGAN EVALUASI —', TRACE_SCENARIO, TRACE_TRAIT)
traces = final_report.get('metric_traces', {})
if TRACE_SCENARIO not in traces:
    print('Belum ada pasangan skor guru dan prediksi test dari model terlatih yang sah.')
    print('Urutan: empat skor guru → verifikasi teks/rubrik/group → training → prediksi test → empat metrik.')
    print('Tidak mengisi langkah perhitungan dengan skor AI atau output head UNTRAINED.')
else:
    trace = traces[TRACE_SCENARIO][TRACE_TRAIT]
    print('1. Cocokkan ID: acuan AI dan prediksi validation pada esai yang sama.' if ai_scope else '1. Cocokkan ID: skor guru dan prediksi test pada esai yang sama.')
    reference_column = 'reference_AI' if ai_scope else 'teacher'
    pair_rows = [{**pair, reference_column:pair['teacher']} for pair in trace['pairs']]
    show_table(['essay_id', reference_column, 'raw_prediction', 'error', 'squared_error',
                'clipped_prediction', 'ordinal_prediction', 'out_of_range'], pair_rows)
    print('2. RMSE: selisih kontinu → kuadrat → mean → akar. Dalam unit skor 1–5.')
    print(json.dumps(trace['RMSE_steps'], ensure_ascii=False, indent=2))
    print('3. Pearson: kurangi mean masing-masing kolom; jumlah produk deviasi dibagi akar produk sum kuadrat.')
    pearson_steps = {key.replace('teacher', 'reference_AI') if ai_scope else key:value for key,value in trace['Pearson_steps'].items()}
    print(json.dumps(pearson_steps, ensure_ascii=False, indent=2))
    print('4. QWK/AC2 menggunakan kategori salinan, bukan membulatkan prediksi untuk RMSE/Pearson.')
    print('Kategori JSON:', trace['categories'], '| Kebijakan midpoint:', trace['midpoint_policy'])
    qwk = trace['QWK_steps']; ac2 = trace['AC2_steps']
    print('5. QWK = 1 − observed weighted disagreement / expected weighted disagreement.')
    print('Observed disagreement:', qwk['observed_weighted_disagreement'],
          '| Expected disagreement:', qwk['expected_weighted_disagreement'], '| QWK:', qwk['QWK'])
    print('6. AC2 = (Pa − Pe)/(1 − Pe), memakai bobot KESEPAKATAN kuadratik.')
    print('Proporsi kategori gabungan:', ac2['pooled_category_proportions'])
    print('Pa:', ac2['Pa_observed_agreement'], '| Pe:', ac2['Pe_chance_agreement'], '| AC2:', ac2['AC2'])
    print('Jumlah kategori selalu lima, termasuk kategori yang tidak muncul dalam subset.')
    print('Metrik undefined:', trace['metrics']['undefined_reasons'])
    def matrix_view(title, matrix):
        print(title, '— baris kategori acuan AI, kolom prediksi' if ai_scope else '— baris kategori guru, kolom kategori prediksi')
        row_label = 'Kategori acuan AI' if ai_scope else 'Kategori guru'
        columns = [row_label] + ['Pred '+str(c) for c in trace['categories']]
        rows = [{row_label:cat, **{'Pred '+str(other):matrix[i][j]
                 for j,other in enumerate(trace['categories'])}} for i,cat in enumerate(trace['categories'])]
        show_table(columns, rows)
    matrix_view('Matriks hitungan pasangan', qwk['observed_counts_teacher_rows_prediction_columns'])
    matrix_view('Hitungan harapan QWK', qwk['expected_counts_from_marginals'])
    matrix_view('Bobot disagreement QWK', qwk['quadratic_disagreement_weights'])
    matrix_view('Bobot agreement AC2', ac2['quadratic_agreement_weights'])
    print('Pasangan validation latihan dengan acuan AI yang sama dengan tabel; bukan akurasi terhadap guru.' if ai_scope else 'Langkah di atas berasal dari pasangan test nyata yang sama dengan tabel hasil, bukan data latihan.')

if ai_scope:
    print('Train/validation:', final_report['manifest']['train'], final_report['manifest']['validation'])
    print('Group:', final_report['manifest']['group_mapping_status'])
    print('Dikecualikan dari latihan:', final_report['manifest']['excluded'])
    show_table(['Skenario','MSE train sebelum','MSE train sesudah','Epoch','Step LR aktif','Sampel parameter berubah'],
               [{'Skenario':s,'MSE train sebelum':info['initial_train_mse'],'MSE train sesudah':info['final_train_mse'],
                 'Epoch':info['completed_epochs'],'Step LR aktif':info['steps_with_nonzero_learning_rate'],
                 'Sampel parameter berubah':info['parameter_sample_changes']} for s,info in final_report['training'].items()])
    print('Checkpoint ini hanya latihan acuan AI; CLI penilaian siswa menolaknya.')
    print('Hasil lama dipakai ulang:', final_report['reused_completed_run'])
```
